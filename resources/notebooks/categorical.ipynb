## 最小可运行起始示例

下面只演示局部操作，不包含完整练习答案。

```python
campaigns = pd.read_csv(DATA_DIR / "campaigns.csv")
by_channel = campaigns.groupby("channel", as_index=False)[["visits", "conversions"]].sum()
# 最小示例只展示访问量；转化率与协调视图留作练习。
if plt is not None:
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(by_channel["channel"], by_channel["visits"])
    ax.set_ylabel("访问会话次数")
    ax.set_title("各渠道访问规模（教学模拟）")
    fig.tight_layout()
    plt.show()
by_channel
```
