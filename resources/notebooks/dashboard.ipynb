## 最小可运行起始示例

下面只演示局部操作，不包含完整练习答案。

```python
retail = pd.read_csv(DATA_DIR / "retail_sales.csv")
retail["order_date"] = pd.to_datetime(retail["order_date"], format="%Y-%m-%d")
retail["sales_net_cny"] = retail["sales_gross_cny"] - retail["refund_cny"]

def filter_retail(frame, regions=None):
    if regions is None:
        return frame.copy()
    return frame.loc[frame["region"].isin(regions)].copy()

selected = filter_retail(retail, regions=["模拟东区"])
assert len(filter_retail(retail)) == len(retail)
assert filter_retail(retail, regions=[]).empty
selected[["quantity", "sales_gross_cny", "refund_cny", "sales_net_cny"]].sum().to_frame("selected_total")
```
