## 最小可运行起始示例

下面只演示局部操作，不包含完整练习答案。

```python
html = (DATA_DIR / "sample_catalog.html").read_text(encoding="utf-8")
try:
    from bs4 import BeautifulSoup
except ModuleNotFoundError:
    BeautifulSoup = None

if BeautifulSoup is not None:
    soup = BeautifulSoup(html, "html.parser")
    first_product = soup.select_one("article.product")
    first_example = {"product_id": first_product.get("data-id"),
                     "name": first_product.select_one("h2").get_text(strip=True)}
else:
    # 标准库预览保证离线基础示例仍可运行；完整字段提取留作练习。
    from html.parser import HTMLParser
    class HeadingPreview(HTMLParser):
        def __init__(self):
            super().__init__()
            self.in_h2 = False
            self.names = []
        def handle_starttag(self, tag, attrs):
            if tag == "h2": self.in_h2 = True
        def handle_endtag(self, tag):
            if tag == "h2": self.in_h2 = False
        def handle_data(self, data):
            if self.in_h2 and data.strip(): self.names.append(data.strip())
    parser = HeadingPreview()
    parser.feed(html)
    first_example = {"name": parser.names[0], "note": "完整练习可选用BeautifulSoup或标准库"}
pd.DataFrame([first_example])
```
